### TFData APTOS Full, IDRID Full, Messidor Full

```python
train_df, val_df = train_test_split(
    aptos_csv,
    test_size=0.2,
    random_state=42,
    stratify=aptos_csv["label_num"]
)

print("Train size      :", len(train_df))
print("Validation size :", len(val_df))
```

```python
VIT_PRESET = "vit_base_patch16_224_imagenet"  # cek nama preset persis di halaman model keras_hub

vit_image_converter = keras_hub.layers.ViTImageConverter.from_preset(VIT_PRESET)

def load_png_vit(image_path, label):
    image = tf.io.read_file(image_path)
    image = tf.image.decode_png(image, channels=3)
    image = vit_image_converter(image)   # otomatis resize + normalize sesuai preset
    return image, label
```

```python
IMG_SIZE = 224
BATCH_SIZE = 32
```

```python
train_dataset = tf.data.Dataset.from_tensor_slices(
    (train_df["image_path"].values, train_df["label_num"].values)
)
train_dataset = train_dataset.map(load_png_vit, num_parallel_calls=AUTOTUNE)
train_dataset = train_dataset.shuffle(len(train_df))
train_dataset = train_dataset.batch(BATCH_SIZE)
train_dataset = train_dataset.prefetch(AUTOTUNE)

val_dataset = tf.data.Dataset.from_tensor_slices(
    (val_df["image_path"].values, val_df["label_num"].values)
)
val_dataset = val_dataset.map(load_png_vit, num_parallel_calls=AUTOTUNE)
val_dataset = val_dataset.batch(BATCH_SIZE)
val_dataset = val_dataset.prefetch(AUTOTUNE)
```

```python
aptos_dataset = tf.data.Dataset.from_tensor_slices(
    (
        aptos_csv["image_path"].values,
        aptos_csv["label_num"].values
    )
)

aptos_dataset = aptos_dataset.map(
    load_png_vit,
    num_parallel_calls=AUTOTUNE
)

aptos_dataset = aptos_dataset.batch(BATCH_SIZE)
aptos_dataset = aptos_dataset.prefetch(AUTOTUNE)

idrid_dataset = tf.data.Dataset.from_tensor_slices(
    (
        idrid_csv["image_path"].values,
        idrid_csv["label_num"].values
    )
)

idrid_dataset = idrid_dataset.map(
    load_png_vit,
    num_parallel_calls=AUTOTUNE
)

idrid_dataset = idrid_dataset.batch(BATCH_SIZE)
idrid_dataset = idrid_dataset.prefetch(AUTOTUNE)

messidor_dataset = tf.data.Dataset.from_tensor_slices(
    (
        messidor_csv["image_path"].values,
        messidor_csv["label_num"].values
    )
)

messidor_dataset = messidor_dataset.map(
    load_png_vit,
    num_parallel_calls=AUTOTUNE
)

messidor_dataset = messidor_dataset.batch(BATCH_SIZE)
messidor_dataset = messidor_dataset.prefetch(AUTOTUNE)
```

### Training

```python
backbone = keras_hub.models.ViTBackbone.from_preset(VIT_PRESET)
backbone.trainable = True

inputs = layers.Input(shape=(IMG_SIZE, IMG_SIZE, 3))
x = backbone(inputs)          # output: sequence token, biasanya (batch, num_patches+1, hidden_dim)
x = x[:, 0, :]                 # ambil CLS token sebagai representasi gambar
x = layers.Dropout(0.3)(x)
outputs = layers.Dense(2, activation='softmax')(x)

model = models.Model(inputs, outputs, name="ViTBase_p_softmax")

model.compile(
    optimizer=SGD(learning_rate=1e-4, momentum=0.9, nesterov=True),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
model.summary()
```

```python
def build_callbacks(
    save_path: str,
    backup_dir: str,
    log_path: str,
    monitor: str = "val_loss",
    patience: int = 10,
):

    os.makedirs(os.path.dirname(save_path), exist_ok=True)
    os.makedirs(backup_dir, exist_ok=True)
    os.makedirs(os.path.dirname(log_path), exist_ok=True)

    checkpoint = ModelCheckpoint(
        filepath=save_path,
        monitor=monitor,
        mode="auto",
        save_best_only=True,
        save_weights_only=False,
        verbose=1,
    )

    early_stop = EarlyStopping(
        monitor=monitor,
        mode="auto",
        patience=patience,
        min_delta=1e-4,
        restore_best_weights=True,
        verbose=1,
    )

    reduce_lr = ReduceLROnPlateau(
        monitor=monitor,
        mode="auto",
        factor=0.5,
        patience=4,
        min_delta=1e-4,
        min_lr=1e-6,
        verbose=1,
    )

    backup = BackupAndRestore(
        backup_dir=backup_dir
    )

    csv_logger = CSVLogger(
        filename=log_path,
        append=True
    )

    return [ checkpoint, early_stop, reduce_lr, backup, csv_logger]
```

```python
callbacks = build_callbacks(
    save_path="/content/drive/MyDrive/Diabetic-Retinopathy/V2/Model/(3)vitbase_p_softmax.keras",
    backup_dir="/content/drive/MyDrive/Diabetic-Retinopathy/V2/Backup/(3)vitbase_p_softmax",
    log_path="/content/drive/MyDrive/Diabetic-Retinopathy/V2/Logs/(3)vitbase_p_softmax.csv",
    monitor="val_loss",
    patience=10,
)

history = model.fit(
    train_dataset,
    validation_data=val_dataset,
    epochs=100,
    callbacks=callbacks
)
```

```python
log = pd.read_csv(
    "/content/drive/MyDrive/Diabetic-Retinopathy/V2/Logs/(3)vitbase_p_softmax.csv"
)

plt.figure(figsize=(10, 5))

plt.plot(log["epoch"] + 1, log["loss"], label="Train Loss")
plt.plot(log["epoch"] + 1, log["val_loss"], label="Validation Loss")

plt.title("Train vs Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.plot(log["epoch"] + 1, log["accuracy"], label="Train Accuracy")
plt.plot(log["epoch"] + 1, log["val_accuracy"], label="Validation Accuracy")

plt.title("Train vs Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)

plt.show()
```

```python
train_loss, train_acc = model.evaluate(train_dataset)
val_loss, val_acc = model.evaluate(val_dataset)

print("PERFORMANCE MODEL (TRAIN VAL)")

print(f"Train Accuracy      : {train_acc:.4f}")
print(f"Validation Accuracy : {val_acc:.4f}")

print(f"\nTrain Loss          : {train_loss:.4f}")
print(f"Validation Loss     : {val_loss:.4f}")
```

### Evaluation on APTOS

```python
y_pred_prob = model.predict(aptos_dataset)

y_pred = np.argmax(y_pred_prob, axis=1)

# true labels
y_true = aptos_csv["label_num"].values
```

```python
print("Classification Report:")
print(classification_report(y_true, y_pred, target_names=["No DR", "DR"], digits=4))

cm = confusion_matrix(y_true, y_pred)
print("\nConfusion Matrix:")
print(cm)
```

```python
# auc scores and roc curve
y_score = y_pred_prob[:, 1]
fpr, tpr, thresholds = roc_curve(y_true, y_score)
auc_score = roc_auc_score(y_true, y_score)
print(f"ROC-AUC: {auc_score:.4f}")
```

```python
plt.figure(figsize=(6,6))
plt.plot(fpr, tpr, label=f"AUC = {auc_score:.4f}")

plt.plot([0,1], [0,1], linestyle="--")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve APTOS Dataset")
plt.legend()

plt.show()
```
